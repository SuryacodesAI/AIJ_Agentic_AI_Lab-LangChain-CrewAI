```python
from crewai import Agent, Task, Crew
from crewai.tools import BaseTool
from duckduckgo_search import DDGS
```

```python
class DuckDuckGoSearchTool(BaseTool):
    name: str = "DuckDuckGo Search"
    description: str = "Search the web for recent information using DuckDuckGo"

    def _run(self, query: str) -> str:
        results = []
        with DDGS() as ddgs:
            for r in ddgs.text(query, max_results=5):
                results.append(r["body"])
        return "\n".join(results)
```

```python
search_tool = DuckDuckGoSearchTool()
```

```python
researcher = Agent(
    role="Web Researcher",
    goal="Find latest and accurate information using web search",
    backstory="Expert in online research and fact finding",
    tools=[search_tool],
    verbose=True
)
```

```python
analyst = Agent(
    role="AI Analyst",
    goal="Analyze research results and extract insights",
    backstory="Expert in pattern recognition",
    verbose=True
)
```

```python
writer = Agent(
    role="Technical Writer",
    goal="Write a clear summary from the analysis",
    backstory="Expert at simplifying complex topics",
    verbose=True
)
```

```python
research_task = Task(
    description="Research latest trends in Generative AI",
    expected_output="Bullet points of AI trends",
    agent=researcher
)
```

```python
analysis_task = Task(
    description="Analyze findings and extract top insights",
    expected_output="Top 5 insights",
    agent=analyst
)
```

```python
writing_task = Task(
    description="Write a 200-word summary based on insights",
    expected_output="Well-written AI trends summary",
    agent=writer
)
```

```python
crew = Crew(
    agents=[researcher, analyst, writer],
    tasks=[research_task, analysis_task, writing_task],
    verbose=True
)
```

```python
result = crew.kickoff()
print("\n===== FINAL OUTPUT =====\n")
print(result)
```
